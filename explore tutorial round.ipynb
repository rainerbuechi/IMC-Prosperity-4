## Plotting the assets

```python
import pandas as pd
from pathlib import Path

DATA_FOLDER = Path("../data/TUTORIAL_ROUND_1")

prices_m1 = pd.read_csv(DATA_FOLDER / "prices_round_0_day_-1.csv", sep=";")
prices_m2 = pd.read_csv(DATA_FOLDER / "prices_round_0_day_-2.csv", sep=";")

trades_m1 = pd.read_csv(DATA_FOLDER / "trades_round_0_day_-1.csv", sep=";")
trades_m2 = pd.read_csv(DATA_FOLDER / "trades_round_0_day_-2.csv", sep=";")
```

```python
prices = pd.concat([prices_m1, prices_m2], ignore_index=True)
trades = pd.concat([trades_m1, trades_m2], ignore_index=True)

prices = prices.sort_values(["day", "timestamp", "product"]).reset_index(drop=True)
trades = trades.sort_values(["timestamp", "symbol"]).reset_index(drop=True)

print(prices.shape)
print(trades.shape)

prices.head()
```

```python
emeralds = prices[prices["product"] == "EMERALDS"].copy().reset_index(drop=True)
tomatoes = prices[prices["product"] == "TOMATOES"].copy().reset_index(drop=True)

emerald_trades = trades[trades["symbol"] == "EMERALDS"].copy().reset_index(drop=True)
tomato_trades = trades[trades["symbol"] == "TOMATOES"].copy().reset_index(drop=True)

print("Emerald rows:", len(emeralds))
print("Tomato rows:", len(tomatoes))
print("Emerald trades:", len(emerald_trades))
print("Tomato trades:", len(tomato_trades))
```

```python
def add_features(df):
    df = df.copy()

    df["spread_1"] = df["ask_price_1"] - df["bid_price_1"]
    df["mid_from_book"] = (df["bid_price_1"] + df["ask_price_1"]) / 2

    bid_num = (
        df["bid_price_1"] * df["bid_volume_1"]
        + df["bid_price_2"] * df["bid_volume_2"]
        + df["bid_price_3"].fillna(0) * df["bid_volume_3"].fillna(0)
    )
    bid_den = (
        df["bid_volume_1"]
        + df["bid_volume_2"]
        + df["bid_volume_3"].fillna(0)
    )

    ask_num = (
        df["ask_price_1"] * df["ask_volume_1"]
        + df["ask_price_2"] * df["ask_volume_2"]
        + df["ask_price_3"].fillna(0) * df["ask_volume_3"].fillna(0)
    )
    ask_den = (
        df["ask_volume_1"]
        + df["ask_volume_2"]
        + df["ask_volume_3"].fillna(0)
    )

    df["weighted_bid"] = bid_num / bid_den
    df["weighted_ask"] = ask_num / ask_den
    df["microprice_like"] = (
        df["ask_price_1"] * df["bid_volume_1"] + df["bid_price_1"] * df["ask_volume_1"]
    ) / (df["bid_volume_1"] + df["ask_volume_1"])

    df["book_imbalance_1"] = (
        df["bid_volume_1"] - df["ask_volume_1"]
    ) / (df["bid_volume_1"] + df["ask_volume_1"])

    return df

emeralds = add_features(emeralds)
tomatoes = add_features(tomatoes)

emeralds.head()
```

```python
summary_cols = [
    "bid_price_1", "ask_price_1", "spread_1", "mid_price",
    "mid_from_book", "microprice_like", "book_imbalance_1"
]

print("EMERALDS")
display(emeralds[summary_cols].describe())

print("TOMATOES")
display(tomatoes[summary_cols].describe())
```

```python
def plot_book_window(df, product_name, start_ts, end_ts):
    sub = df[(df["timestamp"] >= start_ts) & (df["timestamp"] <= end_ts)].copy()

    fig, ax = plt.subplots(figsize=(14, 5))
    ax.plot(sub["timestamp"], sub["bid_price_1"], label="bid 1")
    ax.plot(sub["timestamp"], sub["ask_price_1"], label="ask 1")
    ax.plot(sub["timestamp"], sub["mid_price"], label="mid", linewidth=2)

    if "bid_price_2" in sub.columns:
        ax.plot(sub["timestamp"], sub["bid_price_2"], label="bid 2", alpha=0.5)
        ax.plot(sub["timestamp"], sub["ask_price_2"], label="ask 2", alpha=0.5)

    ax.set_title(f"{product_name} top-of-book window: {start_ts} to {end_ts}")
    ax.set_xlabel("Timestamp")
    ax.set_ylabel("Price")
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.show()
```

```python
plot_book_window(emeralds, "EMERALDS", 0, 20000)
plot_book_window(tomatoes, "TOMATOES", 20000, 40000)
```

```python
def quick_market_stats(df, name):
    out = {
        "product": name,
        "rows": len(df),
        "mid_mean": df["mid_price"].mean(),
        "mid_std": df["mid_price"].std(),
        "spread_mean": df["spread_1"].mean(),
        "spread_std": df["spread_1"].std(),
        "bid_vol1_mean": df["bid_volume_1"].mean(),
        "ask_vol1_mean": df["ask_volume_1"].mean(),
        "imbalance_mean": df["book_imbalance_1"].mean(),
        "return_std": df["mid_return"].std(),
    }
    return pd.DataFrame([out])

stats = pd.concat([
    quick_market_stats(emeralds, "EMERALDS"),
    quick_market_stats(tomatoes, "TOMATOES"),
], ignore_index=True)

stats
```
